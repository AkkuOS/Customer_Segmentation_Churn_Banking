# Customer Segmentation & Churn Pattern Analytics in European Banking

## Notebook 02: Data Cleaning and Preparation

### Objective

This notebook prepares the banking dataset for exploratory analysis, customer segmentation, churn analytics, machine learning, and Streamlit dashboard development.

The main tasks are:

- Reload the raw dataset
- Remove non-analytical fields
- Standardise column formats
- Validate categorical and binary variables
- Inspect unusual numerical values
- Create a clean master dataset
- Preserve customer identifiers separately for dashboard drill-down

```python
import pandas as pd
import numpy as np

from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 140)
pd.set_option("display.float_format", lambda value: f"{value:,.2f}")

print("Libraries imported successfully.")
```

```python
current_directory = Path.cwd()

if current_directory.name.lower() == "notebooks":
    project_root = current_directory.parent
else:
    project_root = current_directory

data_directory = project_root / "data"
output_directory = project_root / "outputs"
table_directory = output_directory / "tables"

table_directory.mkdir(parents=True, exist_ok=True)

raw_dataset_path = data_directory / "Churn_Modelling.csv"
clean_dataset_path = data_directory / "bank_customer_churn_clean.csv"
analytics_dataset_path = data_directory / "bank_customer_churn_analytics.csv"

print("Project root          :", project_root)
print("Raw dataset path      :", raw_dataset_path)
print("Clean dataset path    :", clean_dataset_path)
print("Analytics dataset path:", analytics_dataset_path)
```

```python
if not raw_dataset_path.exists():
    raise FileNotFoundError(
        f"Dataset not found at: {raw_dataset_path}"
    )

df_raw = pd.read_csv(raw_dataset_path)

print("Raw dataset loaded successfully.")
print(f"Rows    : {df_raw.shape[0]:,}")
print(f"Columns : {df_raw.shape[1]:,}")
```

```python
df_raw.head()
```

```python
df_analytics = df_raw.copy()
df_clean = df_raw.copy()

print("Working copies created successfully.")
```

```python
print("Original column names:\n")

for number, column in enumerate(df_raw.columns, start=1):
    print(f"{number:02d}. {column}")
```

```python
column_name_check = pd.DataFrame({
    "Original_Column": df_raw.columns,
    "Stripped_Column": df_raw.columns.str.strip(),
    "Contains_Extra_Spaces": [
        original != stripped
        for original, stripped in zip(
            df_raw.columns,
            df_raw.columns.str.strip()
        )
    ]
})

column_name_check
```

```python
df_analytics.columns = df_analytics.columns.str.strip()
df_clean.columns = df_clean.columns.str.strip()

print("Column names standardised.")
```

```python
print(df_clean.columns.tolist())
```

```python
if "Year" in df_raw.columns:
    print("Unique Year values:", df_raw["Year"].unique())
    print("Number of unique Year values:", df_raw["Year"].nunique())
else:
    print("Year column not found.")
```

```python
if "Year" in df_analytics.columns:
    df_analytics = df_analytics.drop(columns=["Year"])

if "Year" in df_clean.columns:
    df_clean = df_clean.drop(columns=["Year"])

print("Year column removed from both datasets.")
```

```python
if "Surname" in df_analytics.columns:
    df_analytics = df_analytics.drop(columns=["Surname"])

if "Surname" in df_clean.columns:
    df_clean = df_clean.drop(columns=["Surname"])

print("Surname column removed.")
```

```python
print(
    "CustomerId retained in analytics dataset:",
    "CustomerId" in df_analytics.columns
)
```

```python
if "CustomerId" in df_clean.columns:
    df_clean = df_clean.drop(columns=["CustomerId"])

print(
    "CustomerId removed from clean modelling dataset:",
    "CustomerId" not in df_clean.columns
)
```

```python
print("Analytics dataset shape:", df_analytics.shape)
print("Clean dataset shape    :", df_clean.shape)

print("\nAnalytics columns:")
print(df_analytics.columns.tolist())

print("\nClean columns:")
print(df_clean.columns.tolist())
```

```python
text_columns = ["Geography", "Gender"]

for column in text_columns:
    if column in df_analytics.columns:
        df_analytics[column] = (
            df_analytics[column]
            .astype("string")
            .str.strip()
            .str.title()
        )

    if column in df_clean.columns:
        df_clean[column] = (
            df_clean[column]
            .astype("string")
            .str.strip()
            .str.title()
        )

print("Categorical text columns standardised.")
```

```python
for column in text_columns:
    if column in df_clean.columns:
        print(f"\nUnique values in {column}:")
        print(sorted(df_clean[column].dropna().unique().tolist()))
```

```python
binary_columns = [
    "HasCrCard",
    "IsActiveMember",
    "Exited"
]

for column in binary_columns:
    if column in df_analytics.columns:
        df_analytics[column] = pd.to_numeric(
            df_analytics[column],
            errors="coerce"
        ).astype("Int64")

    if column in df_clean.columns:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        ).astype("Int64")

print("Binary columns converted to integer format.")
```

```python
df_clean[binary_columns].dtypes
```

```python
integer_columns = [
    "CreditScore",
    "Age",
    "Tenure",
    "NumOfProducts"
]

float_columns = [
    "Balance",
    "EstimatedSalary"
]

for column in integer_columns:
    if column in df_analytics.columns:
        df_analytics[column] = pd.to_numeric(
            df_analytics[column],
            errors="coerce"
        ).astype("Int64")

    if column in df_clean.columns:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        ).astype("Int64")

for column in float_columns:
    if column in df_analytics.columns:
        df_analytics[column] = pd.to_numeric(
            df_analytics[column],
            errors="coerce"
        )

    if column in df_clean.columns:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        )

print("Numerical columns converted successfully.")
```

```python
missing_after_conversion = pd.DataFrame({
    "Column": df_clean.columns,
    "Missing_Count": df_clean.isna().sum().values,
    "Missing_Percentage": (
        df_clean.isna().mean().values * 100
    ).round(2)
})

missing_after_conversion
```

```python
total_missing_after_conversion = int(
    df_clean.isna().sum().sum()
)

print(
    f"Total missing values after conversion: "
    f"{total_missing_after_conversion:,}"
)
```

```python
analytics_duplicate_rows = int(
    df_analytics.duplicated().sum()
)

clean_duplicate_rows = int(
    df_clean.duplicated().sum()
)

print(
    f"Duplicate rows in analytics dataset: "
    f"{analytics_duplicate_rows:,}"
)

print(
    f"Duplicate rows in clean dataset: "
    f"{clean_duplicate_rows:,}"
)
```

```python
duplicate_customer_ids = int(
    df_analytics["CustomerId"].duplicated().sum()
)

print(
    f"Duplicate CustomerId values: "
    f"{duplicate_customer_ids:,}"
)
```

```python
expected_categories = {
    "Geography": {"France", "Germany", "Spain"},
    "Gender": {"Female", "Male"}
}

categorical_validation_results = []

for column, expected_values in expected_categories.items():
    actual_values = set(
        df_clean[column].dropna().unique()
    )

    unexpected_values = actual_values - expected_values
    missing_expected_values = expected_values - actual_values

    categorical_validation_results.append({
        "Column": column,
        "Actual_Values": str(sorted(actual_values)),
        "Unexpected_Values": str(sorted(unexpected_values)),
        "Missing_Expected_Values": str(
            sorted(missing_expected_values)
        ),
        "Validation_Passed": len(unexpected_values) == 0
    })

categorical_validation_df = pd.DataFrame(
    categorical_validation_results
)

categorical_validation_df
```

```python
binary_validation_results = []

for column in binary_columns:
    actual_values = set(
        df_clean[column].dropna().unique()
    )

    invalid_values = actual_values - {0, 1}

    binary_validation_results.append({
        "Column": column,
        "Actual_Values": str(sorted(actual_values)),
        "Invalid_Values": str(sorted(invalid_values)),
        "Validation_Passed": len(invalid_values) == 0
    })

binary_validation_df = pd.DataFrame(
    binary_validation_results
)

binary_validation_df
```

```python
range_rules = {
    "CreditScore": {
        "minimum": 300,
        "maximum": 900
    },
    "Age": {
        "minimum": 18,
        "maximum": 100
    },
    "Tenure": {
        "minimum": 0,
        "maximum": 10
    },
    "Balance": {
        "minimum": 0,
        "maximum": None
    },
    "NumOfProducts": {
        "minimum": 1,
        "maximum": 4
    },
    "EstimatedSalary": {
        "minimum": 0,
        "maximum": None
    }
}

range_validation_results = []

for column, limits in range_rules.items():
    minimum_allowed = limits["minimum"]
    maximum_allowed = limits["maximum"]

    below_minimum = int(
        (df_clean[column] < minimum_allowed).sum()
    )

    if maximum_allowed is not None:
        above_maximum = int(
            (df_clean[column] > maximum_allowed).sum()
        )
    else:
        above_maximum = 0

    range_validation_results.append({
        "Column": column,
        "Observed_Minimum": df_clean[column].min(),
        "Observed_Maximum": df_clean[column].max(),
        "Below_Minimum_Count": below_minimum,
        "Above_Maximum_Count": above_maximum,
        "Validation_Passed": (
            below_minimum == 0 and above_maximum == 0
        )
    })

range_validation_df = pd.DataFrame(
    range_validation_results
)

range_validation_df
```

```python
unusual_records = df_analytics[
    (df_analytics["EstimatedSalary"] < 1000)
    | (df_analytics["Age"] >= 80)
    | (df_analytics["Balance"] >= 200000)
].copy()

print(
    f"Number of records flagged for inspection: "
    f"{len(unusual_records):,}"
)

unusual_records.head(20)
```

```python
zero_balance_count = int(
    (df_clean["Balance"] == 0).sum()
)

zero_balance_percentage = (
    zero_balance_count / len(df_clean) * 100
)

print(f"Zero-balance customers: {zero_balance_count:,}")
print(
    f"Zero-balance percentage: "
    f"{zero_balance_percentage:.2f}%"
)
```

```python
zero_balance_churn = (
    df_clean
    .assign(
        Balance_Status=np.where(
            df_clean["Balance"] == 0,
            "Zero Balance",
            "Positive Balance"
        )
    )
    .groupby("Balance_Status")
    .agg(
        Customers=("Exited", "size"),
        Churned_Customers=("Exited", "sum"),
        Churn_Rate=("Exited", "mean")
    )
    .reset_index()
)

zero_balance_churn["Churn_Rate"] = (
    zero_balance_churn["Churn_Rate"] * 100
).round(2)

zero_balance_churn
```

```python
df_analytics["Churn_Status"] = df_analytics["Exited"].map({
    0: "Retained",
    1: "Churned"
})

df_analytics["Credit_Card_Status"] = (
    df_analytics["HasCrCard"].map({
        0: "No Credit Card",
        1: "Has Credit Card"
    })
)

df_analytics["Activity_Status"] = (
    df_analytics["IsActiveMember"].map({
        0: "Inactive",
        1: "Active"
    })
)

df_clean["Churn_Status"] = df_clean["Exited"].map({
    0: "Retained",
    1: "Churned"
})

df_clean["Credit_Card_Status"] = (
    df_clean["HasCrCard"].map({
        0: "No Credit Card",
        1: "Has Credit Card"
    })
)

df_clean["Activity_Status"] = (
    df_clean["IsActiveMember"].map({
        0: "Inactive",
        1: "Active"
    })
)

print("Readable label columns created.")
```

```python
analytics_column_order = [
    "CustomerId",
    "CreditScore",
    "Geography",
    "Gender",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "Credit_Card_Status",
    "IsActiveMember",
    "Activity_Status",
    "EstimatedSalary",
    "Exited",
    "Churn_Status"
]

df_analytics = df_analytics[
    analytics_column_order
]
```

```python
clean_column_order = [
    "CreditScore",
    "Geography",
    "Gender",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "Credit_Card_Status",
    "IsActiveMember",
    "Activity_Status",
    "EstimatedSalary",
    "Exited",
    "Churn_Status"
]

df_clean = df_clean[
    clean_column_order
]
```

## Data Cleaning and Preparation Findings

- The raw dataset contained **10,000 customer records and 14 columns**.
- The `Year` column was removed because it contained only the constant value `2025`.
- The `Surname` field was removed because it does not provide useful analytical or predictive information.
- `CustomerId` was retained in the analytics dataset for customer search and drill-down.
- `CustomerId` was removed from the modelling-ready dataset to prevent identifier leakage.
- Geography and gender values were standardised.
- Binary variables were validated to contain only `0` and `1`.
- Numerical fields were converted to appropriate data types.
- No missing values were introduced during type conversion.
- No duplicate customer IDs were identified.
- Zero account balance was retained as a valid customer characteristic rather than treated as missing data.
- Unusual numerical values were flagged for later exploration but were not removed without evidence.
- Readable labels were created for churn, activity, and credit-card status.
- Two datasets were generated:
  - `bank_customer_churn_analytics.csv` for EDA and dashboard development
  - `bank_customer_churn_clean.csv` for feature engineering and machine learning preparation
