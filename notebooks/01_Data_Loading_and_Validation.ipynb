### Project Objective

```python
binary_columns = [
    "HasCrCard",
    "IsActiveMember",
    "Exited"
]

binary_validation_results = []

for column in binary_columns:
    if column in df.columns:
        unique_values = sorted(
            df[column].dropna().unique().tolist()
        )

        valid_binary = set(unique_values).issubset({0, 1})

        binary_validation_results.append(
            {
                "Column": column,
                "Unique_Values": str(unique_values),
                "Valid_Binary_Encoding": valid_binary
            }
        )

binary_validation_df = pd.DataFrame(
    binary_validation_results
)

binary_validation_df
```

```python
if not binary_validation_df.empty:
    if binary_validation_df[
        "Valid_Binary_Encoding"
    ].all():
        print(
            "Validation passed: All available binary columns "
            "contain only 0 and 1."
        )
    else:
        print(
            "Validation warning: At least one binary column "
            "contains unexpected values."
        )
```

```python
categorical_value_checks = {}

for column in ["Geography", "Gender"]:
    if column in df.columns:
        categorical_value_checks[column] = (
            df[column]
            .value_counts(dropna=False)
            .sort_index()
        )

for column, counts in categorical_value_checks.items():
    print(f"\nValue counts for {column}:")
    print(counts)
```

```python
target_column = "Exited"

if target_column not in df.columns:
    raise KeyError(
        f"The required target column '{target_column}' "
        "was not found in the dataset."
    )

target_counts = (
    df[target_column]
    .value_counts(dropna=False)
    .sort_index()
)

target_percentages = (
    df[target_column]
    .value_counts(
        normalize=True,
        dropna=False
    )
    .sort_index()
    .mul(100)
    .round(2)
)

target_distribution = pd.DataFrame(
    {
        "Customer_Count": target_counts,
        "Percentage": target_percentages
    }
)

target_distribution.index.name = "Exited"
target_distribution
```

```python
total_customers = len(df)
retained_customers = int((df["Exited"] == 0).sum())
churned_customers = int((df["Exited"] == 1).sum())
overall_churn_rate = df["Exited"].mean() * 100

print(f"Total customers    : {total_customers:,}")
print(f"Retained customers : {retained_customers:,}")
print(f"Churned customers  : {churned_customers:,}")
print(f"Overall churn rate : {overall_churn_rate:.2f}%")
```

```python
numerical_validation_rules = {
    "CreditScore": (0, 1000),
    "Age": (18, 120),
    "Tenure": (0, None),
    "Balance": (0, None),
    "NumOfProducts": (0, None),
    "EstimatedSalary": (0, None)
}

numerical_validation_results = []

for column, limits in numerical_validation_rules.items():
    if column in df.columns:
        minimum_allowed, maximum_allowed = limits

        invalid_low_count = int(
            (df[column] < minimum_allowed).sum()
        )

        if maximum_allowed is not None:
            invalid_high_count = int(
                (df[column] > maximum_allowed).sum()
            )
        else:
            invalid_high_count = 0

        numerical_validation_results.append(
            {
                "Column": column,
                "Minimum_Observed": df[column].min(),
                "Maximum_Observed": df[column].max(),
                "Invalid_Below_Minimum": invalid_low_count,
                "Invalid_Above_Maximum": invalid_high_count
            }
        )

numerical_validation_df = pd.DataFrame(
    numerical_validation_results
)

numerical_validation_df
```

```python
validation_report = pd.DataFrame(
    {
        "Validation_Check": [
            "Dataset loaded",
            "Dataset contains records",
            "Target column available",
            "Missing values",
            "Duplicate rows",
            "Duplicate customer IDs",
            "Valid target encoding"
        ],
        "Result": [
            True,
            len(df) > 0,
            "Exited" in df.columns,
            total_missing_values == 0,
            duplicate_row_count == 0,
            (
                duplicate_customer_id_count == 0
                if "CustomerId" in df.columns
                else None
            ),
            (
                set(df["Exited"].dropna().unique())
                .issubset({0, 1})
                if "Exited" in df.columns
                else False
            )
        ],
        "Details": [
            dataset_path.name,
            f"{len(df):,} records",
            "Exited" if "Exited" in df.columns else "Not found",
            f"{total_missing_values:,} missing values",
            f"{duplicate_row_count:,} duplicate rows",
            (
                f"{duplicate_customer_id_count:,} duplicate IDs"
                if "CustomerId" in df.columns
                else "CustomerId not available"
            ),
            (
                str(sorted(df["Exited"].dropna().unique().tolist()))
                if "Exited" in df.columns
                else "Target unavailable"
            )
        ]
    }
)

validation_report
```

```python
validation_report_path = (
    table_directory / "initial_validation_report.csv"
)

validation_report.to_csv(
    validation_report_path,
    index=False
)

print(
    f"Validation report saved to: "
    f"{validation_report_path}"
)
```
